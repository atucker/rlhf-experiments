### Baselines

In the RLOO (Reinforce Leave-One-Out estimator), we use a variance-reducing baseline by computing an estimate of the expected return of the policy on a given prompt.

The RLOO loss formulation is as follows:

$$\mathcal{L} = \mathop{\mathbb{E}}_{y^{(1)}...y^{(k)}\sim \pi_\theta(\cdot | x)} \left[\sum_{i=1}^k \left[R(y^{(i)}, x) - \frac{1}{k-1}\sum_{j \neq i} R(y^{(j)}, x) \right] \log \pi_\theta(y|x) \right]$$

Where $R(y, x)$ is the sum of the reward given by the reward model and the KL reward penalty.

When *rloo_k* = 1, we set the baseline to a constant 0. This yields

$$\mathcal{L_{k=1}} = \mathop{\mathbb{E}}_{y^{(1)}...y^{(k)}\sim \pi_\theta(\cdot | x)} \left[\sum_{i=1}^k \left[R(y^{(i)}, x) \right] \log \pi_\theta(y|x)\right]$$

with the gradient 

$$\sum_{i=1}^k \left[R(y^{(i)}, x) \right] \log \pi_\theta(y|x)$$
