```python
import json
from typing import List

def merge_alpaca_eval_outputs(file_paths: List[str], model_names: List[str]) -> List[dict]:
    """
    Merge AlpacaEval outputs from multiple files into a single list of dictionaries.
    len(file_paths) should be equal to len(model_names)
    """
    data = []
    for file_path, model_name in zip(file_paths, model_names):
        with open(file_path, "r") as f:
            generated_outputs = json.load(f)
            for elem in generated_outputs:
                elem["generator"] = model_name
                data.append(elem)
    return data

outputs = merge_alpaca_eval_outputs(
    file_paths=["ultrafeedback_rloo_merged_alpaca_eval_outputs_final.json", "ultrafeedback_rloo_patch_merged_alpaca_eval_outputs_final.json"],
    model_names=["Llama-3.1-8B-RLOO", "Llama-3.1-8B-RLOO-Patch"]
)
```
